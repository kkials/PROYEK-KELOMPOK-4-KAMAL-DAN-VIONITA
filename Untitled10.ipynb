```python
#
import datetime

# Daftar Menu
menu = {
    1: ["Kopi Jago Latte", 15000],
    2: ["Americano", 12000],
    3: ["Kopi Susu Gula Aren", 14000],
    4: ["Matcha Latte", 16000],
    5: ["Roti Bakar", 10000]
}

# Riwayat Transaksi
riwayat = []

# Menampilkan Menu
def tampilkan_menu():
    print("\n=== MENU JAGO COFFEE ===")
    for nomor, data in menu.items():
        print(f"{nomor:<3} {data[0]:<25} - Rp {data[1]:>6,}")

# Transaksi Baru
def transaksi():
    tampilkan_menu()

    total = 0
    pesanan = []

    while True:
        pilih = int(input("\nPilih menu (0 untuk selesai): "))

        if pilih == 0:
            break

        if pilih in menu:
            jumlah = int(input("Jumlah pesanan: "))

            nama = menu[pilih][0]
            harga = menu[pilih][1]
            subtotal = harga * jumlah

            total += subtotal
            pesanan.append(nama)

            print(nama, "x", jumlah, "=", subtotal)

        else:
            print("Menu tidak tersedia!")

    if total == 0:
        print("Transaksi dibatalkan.")
        return

    print("\nTotal pembayaran: Rp", total)

    print("\n1. QRIS")
    print("2. Cash")

    metode = input("Pilih pembayaran: ")

    if metode == "1":
        konfirmasi = input("Pembayaran QRIS berhasil? (y/n): ")

        if konfirmasi.lower() != "y":
            print("Pembayaran gagal.")
            return

        metode_bayar = "QRIS"

    elif metode == "2":
        bayar = int(input("Masukkan uang: Rp "))

        if bayar < total:
            print("Uang tidak cukup!")
            return

        print("Kembalian: Rp", bayar - total)
        metode_bayar = "Cash"

    else:
        print("Metode tidak tersedia.")
        return

    waktu = datetime.datetime.now() + datetime.timedelta(hours=7)

    riwayat.append({
        "Waktu": waktu,
        "Pesanan": ", ".join(pesanan),
        "Metode": metode_bayar,
        "Total": total
    })

    print("\nTransaksi berhasil disimpan!")

# Laporan Penjualan

def laporan():
    print("\n" + "="*70)
    print("             LAPORAN PENJUALAN JAGO COFFEE")
    print("="*70)

    if len(riwayat) == 0:
        print("Belum ada transaksi.")
        return

    print(f"{'WAKTU':<20} {'PESANAN':<25} {'METODE':<10} {'TOTAL':>10}")
    print("-"*70)

    omzet = 0

    for data in riwayat:
        waktu = data["Waktu"].strftime("%H:%M:%S")
        pesanan = data["Pesanan"]
        metode = data["Metode"]
        total = data["Total"]

        print(f"{waktu:<20} {pesanan:<25} {metode:<10} {total:>10,}")

        omzet += total

    print("-"*65)
    print(f"{'TOTAL OMZET':<55} Rp {omzet:,}")
    print(f"{'JUMLAH TRANSAKSI':<55} {len(riwayat)}")
    print("="*65)

# Program Utama
while True:
    print("\n=== POS JAGO COFFEE ===")
    print("1. Transaksi Baru")
    print("2. Laporan Penjualan")
    print("3. Keluar")

    pilihan = input("Pilih menu: ")

    if pilihan == "1":
        transaksi()

    elif pilihan == "2":
        laporan()

    elif pilihan == "3":
        print("Program selesai.")
        break

    else:
        print("Pilihan tidak valid!")
```
