```python
import datetime
import pandas as pd

# Database Menu Jago Coffee
DAFTAR_MENU = {
    1: {"nama": "Kopi Jago Latte", "harga": 15000},
    2: {"nama": "Americano", "harga": 12000},
    3: {"nama": "Kopi Susu Gula Aren", "harga": 14000},
    4: {"nama": "Matcha Latte", "harga": 16000},
    5: {"nama": "Roti Bakar", "harga": 10000}
}

# Database Penyimpanan Transaksi Harian (In-Memory Data)
riwayat_transaksi = []

def tampilkan_menu():
    print("\n" + "="*40)
    print("        MENU KASIR JAGO COFFEE        ")
    print("="*40)
    for id_item, detail in DAFTAR_MENU.items():
        print(f"[{id_item}] {detail['nama']:<22} : Rp {detail['harga']:,}")
    print("="*40)

def transaksi_baru():
    tampilkan_menu()
    pesanan = []
    total_bayar = 0

    print("\n--- POS INPUT PESANAN ---")
    while True:
        try:
            pilihan = int(input("Masukkan nomor menu (0 untuk selesai): "))
            if pilihan == 0:
                break
            if pilihan in DAFTAR_MENU:
                jumlah = int(input(f"Jumlah '{DAFTAR_MENU[pilihan]['nama']}': "))
                subtotal = DAFTAR_MENU[pilihan]['harga'] * jumlah
                total_bayar += subtotal

                pesanan.append({
                    "item": DAFTAR_MENU[pilihan]['nama'],
                    "jumlah": jumlah,
                    "subtotal": subtotal
                })
                print(f"-> Ditambahkan: {DAFTAR_MENU[pilihan]['nama']} x{jumlah} = Rp {subtotal:,}")
            else:
                print("⚠️ Nomor menu tidak valid!")
        except ValueError:
            print("⚠️ Input harus berupa angka!")

    if not pesanan:
        print("Transaksi dibatalkan (tidak ada item dipesan).")
        return

    print("\n" + "-"*40)
    print(f"TOTAL YANG HARUS DIBAYAR: Rp {total_bayar:,}")
    print("-"*40)

    # Verifikasi Pembayaran (Simulasi Real-Time QRIS vs Cash)
    print("Pilih Metode Pembayaran:")
    print("1. QRIS (Auto-Verifikasi API)")
    print("2. Cash / Tunai")

    metode = input("Pilihan (1/2): ")
    status_bayar = False

    if metode == "1":
        print("\n[SYSTEM] Menghubungkan ke Gateway QRIS...")
        print("[SYSTEM] Menunggu Notifikasi Pembayaran Masuk...")
        # Simulasi Verifikasi Pembayaran Real-Time
        konfirmasi = input("Apakah dana QRIS masuk ke rekening penyedia? (y/n): ").lower()
        if konfirmasi == 'y':
            status_bayar = True
            print("✅ NOTIFIKASI: Pembayaran QRIS Sukses Terverifikasi!")
        else:
            print("❌ GAGAL: Pembayaran QRIS Tidak Ditemukan / Transaksi Dibatalkan.")
            return
    elif metode == "2":
        try:
            bayar = int(input("Nominal Uang Tunai: Rp "))
            if bayar >= total_bayar:
                kembalian = bayar - total_bayar
                print(f"✅ Kembalian: Rp {kembalian:,}")
                status_bayar = True
            else:
                print("❌ Uang tunai kurang! Transaksi dibatalkan.")
                return
        except ValueError:
            print("⚠️ Input angka salah.")
            return
    else:
        print("Pilihan metode pembayaran tidak ada.")
        return

    # Simpan Transaksi jika Berhasil
    if status_bayar:
        waktu_sekarang = datetime.datetime.now().strftime("%Y-%m-%d %H:%M:%S")
        riwayat_transaksi.append({
            "Waktu": waktu_sekarang,
            "Detail Pesanan": ", ".join([f"{p['item']} (x{p['jumlah']})" for p in pesanan]),
            "Metode": "QRIS" if metode == "1" else "Cash",
            "Total": total_bayar,
            "Status": "PASS"
        })
        print("\n🎉 Transaksi Berhasil Disimpan Ke Database!")

def tampilkan_laporan_harian():
    print("\n" + "="*50)
    print("      LAPORAN REKAPITULASI PENJUALAN HARIAN      ")
    print("="*50)

    if not riwayat_transaksi:
        print("Belum ada data transaksi untuk hari ini.")
        return

    df = pd.DataFrame(riwayat_transaksi)
    print(df.to_string(index=False))

    total_omzet = df["Total"].sum()
    print("-" * 50)
    print(f"TOTAL OOMZET HARIAN : Rp {total_omzet:,}")
    print(f"TOTAL TRANSAKSI     : {len(df)} Transaksi")
    print("=" * 50)

# Main Loop Aplikasi
def main():
    while True:
        print("\n=== SISTEM POS JAGO COFFEE (VERSI 1.0) ===")
        print("1. Transaksi Baru (Kasir)")
        print("2. Laporan Penjualan Harian (Pemilik)")
        print("3. Keluar")

        pilih = input("Pilih Menu (1-3): ")
        if pilih == "1":
            transaksi_baru()
        elif pilih == "2":
            tampilkan_laporan_harian()
        elif pilih == "3":
            print("Aplikasi ditutup. Terima kasih!")
            break
        else:
            print("Pilihan tidak valid.")

if __name__ == "__main__":
    main()
```
